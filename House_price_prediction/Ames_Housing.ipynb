```python
X = df.iloc[:, 0:81]
Y = df.iloc[:, -1]
```

```python
print(df.shape)
print(X.shape)
print(Y.shape)
```

```python
cat_cols = X.select_dtypes(include='object').columns
num_cols = X.select_dtypes(exclude='object').columns
```

```python
from sklearn.compose import ColumnTransformer ## different preprocessing for differnt columns like - categorical or numerical
from sklearn.preprocessing import OneHotEncoder, StandardScaler ## OneHotEncoder for categorical , and StandardScaler for numerical
from sklearn.impute import SimpleImputer ## handle nan values
from sklearn.pipeline import Pipeline

num_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median')), ## first convert nan value to median
    ('scaler', StandardScaler())  ## then scale all the values
])

cat_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')), ## first convert nan to most frequent data
    ('onehot', OneHotEncoder(handle_unknown='ignore')) ##
])
```

```python
preprocessor = ColumnTransformer( ## now these pipeline is apply on the columns
    transformers=[
        ('num', num_pipeline, num_cols),
        ('cat', cat_pipeline, cat_cols)
    ]
)
```

TRAIN TEST SPLIT

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, Y, test_size=0.2, random_state=42
)
```

```python
X_train = preprocessor.fit_transform(X_train)
X_test = preprocessor.transform(X_test)
```

STANDARIZING

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler(with_mean=False)
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

TRAIN MODEL

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import cross_val_score
model = LinearRegression()
model.fit(X_train, y_train)
```

PREDICTION

```python
X_test_predict=model.predict(X_test)
X_train_predict=model.predict(X_train)
```
